```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def __str__(self):
        status = "Available" if self.is_available else "Borrowed"
        return f"'{self.title}' by {self.author} — [{status}]"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author):
        self.books.append(Book(title, author))
        print(f"  Book '{title}' added successfully.")

    def display_books(self):
        if not self.books:
            print("  No books in the library.")
            return
        print("\n  === Library Books ===")
        for i, book in enumerate(self.books, 1):
            print(f"  {i}. {book}")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if book.is_available:
                    book.is_available = False
                    print(f"  Successfully borrowed '{book.title}'.")
                else:
                    print(f"  Sorry, '{book.title}' is already borrowed.")
                return
        print(f"  Book '{title}' not found in the library.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower():
                if not book.is_available:
                    book.is_available = True
                    print(f"  Successfully returned '{book.title}'.")
                else:
                    print(f"  '{book.title}' was not borrowed.")
                return
        print(f"  Book '{title}' not found in the library.")


def library_system():
    library = Library()
    print("=== Library Book Management System ===\n")

    # Pre-filled books
    prefilled_books = [
        ("Harry Potter and the Sorcerer's Stone", "J.K. Rowling"),
        ("Harry Potter and the Chamber of Secrets", "J.K. Rowling"),
        ("Harry Potter and the Prisoner of Azkaban", "J.K. Rowling"),
        ("Harry Potter and the Goblet of Fire", "J.K. Rowling"),
        ("Harry Potter and the Order of the Phoenix", "J.K. Rowling"),
    ]

    print("Adding pre-loaded books to the library...")
    for title, author in prefilled_books:
        library.add_book(title, author)

    while True:
        print("\n--- Menu ---")
        print("1. View all books")
        print("2. Borrow a book")
        print("3. Return a book")
        print("4. Exit")

        choice = input("Choose an option (1-4): ").strip()

        if choice == "1":
            library.display_books()
        elif choice == "2":
            title = input("  Enter book title to borrow: ").strip()
            library.borrow_book(title)
        elif choice == "3":
            title = input("  Enter book title to return: ").strip()
            library.return_book(title)
        elif choice == "4":
            print("Exiting Library System. Goodbye!")
            break
        else:
            print("  Invalid choice. Please enter 1-4.")

library_system()
```

```python
class Employee:
    OVERTIME_THRESHOLD = 160
    OVERTIME_MULTIPLIER = 1.5

    def __init__(self, name, hourly_wage, hours_worked):
        self.name = name
        self.hourly_wage = hourly_wage
        self.hours_worked = hours_worked

    def compute_salary(self):
        if self.hours_worked <= self.OVERTIME_THRESHOLD:
            base_salary = self.hourly_wage * self.hours_worked
            overtime_pay = 0.0
        else:
            base_salary = self.hourly_wage * self.OVERTIME_THRESHOLD
            extra_hours = self.hours_worked - self.OVERTIME_THRESHOLD
            overtime_pay = extra_hours * self.hourly_wage * self.OVERTIME_MULTIPLIER
        total = base_salary + overtime_pay
        return base_salary, overtime_pay, total

    def display(self):
        base, overtime, total = self.compute_salary()
        print(f"\n  Employee : {self.name}")
        print(f"  Hours    : {self.hours_worked} hrs")
        print(f"  Base Pay : P{base:,.2f}")
        if overtime > 0:
            print(f"  Overtime : P{overtime:,.2f}")
        print(f"  Total    : P{total:,.2f}")
        print(f"  {'─' * 35}")


def payroll_system():
    print("=== Employee Payroll System ===\n")
    employees = []

    while True:
        try:
            n = int(input("Enter number of employees: "))
            if n <= 0:
                print("Please enter at least 1 employee.")
                continue
            break
        except ValueError:
            print("Invalid input. Enter a whole number.")

    for i in range(n):
        print(f"\n  Employee {i + 1}:")

        # Pre-filled name for first employee
        if i == 0:
            name = "Aisy Jibril Cordero"
            print(f"    Name: {name}")
        else:
            name = input("    Name: ").strip()
            if not name:
                print("    Name cannot be empty. Skipping.")
                continue

        while True:
            try:
                wage = float(input("    Hourly wage (P): "))
                if wage < 0:
                    print("    Wage cannot be negative.")
                    continue
                break
            except ValueError:
                print("    Invalid input. Enter a number.")

        while True:
            try:
                hours = float(input("    Hours worked: "))
                if hours < 0:
                    print("    Hours cannot be negative.")
                    continue
                break
            except ValueError:
                print("    Invalid input. Enter a number.")

        employees.append(Employee(name, wage, hours))

    print("\n=== Payroll Summary ===")
    print(f"  {'─' * 35}")
    for emp in employees:
        emp.display()

payroll_system()
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def __str__(self):
        return f"{self.name} — P{self.price:.2f} (Stock: {self.stock})"


class ShoppingCart:
    def __init__(self):
        self.items = {}  # {product_name: (Product, quantity)}

    def add_item(self, product, quantity=1):
        if quantity > product.stock:
            print(f"  Sorry, only {product.stock} unit(s) of '{product.name}' in stock.")
            return
        if product.name in self.items:
            existing_qty = self.items[product.name][1]
            if existing_qty + quantity > product.stock:
                print(f"  Cannot add {quantity} more. Stock limit reached.")
                return
            self.items[product.name] = (product, existing_qty + quantity)
        else:
            self.items[product.name] = (product, quantity)
        product.stock -= quantity
        print(f"  Added {quantity}x '{product.name}' to cart.")

    def remove_item(self, product_name):
        if product_name in self.items:
            product, qty = self.items.pop(product_name)
            product.stock += qty
            print(f"  Removed '{product_name}' from cart.")
        else:
            print(f"  '{product_name}' is not in your cart.")

    def view_cart(self):
        if not self.items:
            print("  Your cart is empty.")
            return
        print("\n  === Your Cart ===")
        total = 0
        for name, (product, qty) in self.items.items():
            subtotal = product.price * qty
            total += subtotal
            print(f"  - {name} x{qty} @ P{product.price:.2f} = P{subtotal:.2f}")
        print(f"  {'─' * 35}")
        print(f"  Total: P{total:.2f}")

    def checkout(self):
        if not self.items:
            print("  Cart is empty. Nothing to checkout.")
            return
        total = sum(p.price * q for p, q in self.items.values())
        print(f"\n  === Checkout ===")
        print(f"  Total Amount Due: P{total:.2f}")
        print("  Thank you for shopping!")
        self.items.clear()


def shopping_system():
    catalog = [
        Product("Laptop",     45000.00, 5),
        Product("Mouse",        350.00, 20),
        Product("Keyboard",    1200.00, 15),
        Product("Monitor",    12000.00, 8),
        Product("USB Hub",      800.00, 10),
    ]
    cart = ShoppingCart()

    print("=== Online Shopping Cart ===\n")

    while True:
        print("\n--- Menu ---")
        print("1. View products")
        print("2. Add item to cart")
        print("3. Remove item from cart")
        print("4. View cart")
        print("5. Checkout")
        print("6. Exit")

        choice = input("Choose an option (1-6): ").strip()

        if choice == "1":
            print("\n  === Product Catalog ===")
            for i, p in enumerate(catalog, 1):
                print(f"  {i}. {p}")

        elif choice == "2":
            print("\n  === Product Catalog ===")
            for i, p in enumerate(catalog, 1):
                print(f"  {i}. {p}")
            try:
                idx = int(input("  Select product number: ")) - 1
                if idx < 0 or idx >= len(catalog):
                    print("  Invalid product number.")
                    continue
                qty = int(input("  Enter quantity: "))
                if qty <= 0:
                    print("  Quantity must be at least 1.")
                    continue
                cart.add_item(catalog[idx], qty)
            except ValueError:
                print("  Invalid input. Enter a number.")

        elif choice == "3":
            name = input("  Enter product name to remove: ").strip()
            cart.remove_item(name)

        elif choice == "4":
            cart.view_cart()

        elif choice == "5":
            cart.checkout()

        elif choice == "6":
            print("Exiting. Goodbye!")
            break

        else:
            print("  Invalid choice. Please enter 1-6.")

shopping_system()
```
