```python
# Step 1: Define a class named Laptop
class Laptop:
    def __init__(self, brand, storage, ram):
        self.brand = brand
        self.model = storage 
        self.year = ram
    
    def display_info(self):
        return f"{self.year} {self.brand} {self.model}"

# Step 2: Create an object of the Laptop class
my_laptop = Laptop("Lenovo LOQ", "512 GB", "12 GB RAM")

# Step 3: Print the laptop details using the method
print(my_laptop.display_info())
```

```python
class StudentRecord:
    def __init__(self):
        self.__grades = []

    def add_grade(self, grade):
        self.__grades.append(grade)

    def get_grades(self):
        return f"Grades: {self.__grades}"

record = StudentRecord()
record.add_grade(88)
record.add_grade(92)
record.add_grade(75)
print(record.get_grades())
```

```python
class Device:
    def __init__(self, device_name):
        self.device_name = device_name

    def power_on(self):
        return f"{self.device_name} is powering on."

class Smartphone(Device):
    def power_on(self):
        return f"{self.device_name} smartphone is booting. Loading apps..."

d = Device("Generic Device")
s = Smartphone("Samsung Galaxy")
print(d.power_on())
print(s.power_on())
```

```python
class Device:
    def power_on(self):
        return "Generic device powering on."

class Smartphone(Device):
    def power_on(self):
        return "Smartphone: booting Android OS..."

class Laptop(Device):
    def power_on(self):
        return "Laptop: loading Windows, please wait..."

def start_device(device):
    print(device.power_on())

start_device(Smartphone())
start_device(Laptop())
```

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32

celsius = float(input("Enter temperature in Celsius: "))
result = TemperatureConverter.celsius_to_fahrenheit(celsius)
print(f"{celsius}°C = {result:.2f}°F")
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(3, 4)
v2 = Vector(1, 2)
print(v1 + v2)
print(v1 - v2)
```

```python
from abc import ABC, abstractmethod

class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

class WashingMachine(Appliance):
    def turn_on(self):
        return "Washing machine is now running."

class Refrigerator(Appliance):
    def turn_on(self):
        return "Refrigerator is now cooling."

wm = WashingMachine()
fridge = Refrigerator()
print(wm.turn_on())
print(fridge.turn_on())
```

```python
class A:
    def show(self):
        return "Class A"

class B(A):
    def show(self):
        return "Class B"

class C(A):
    def show(self):
        return "Class C"

class D(B, C):
    pass

obj = D()
print(obj.show())       # Class B (MRO picks B first)
print(D.mro())          # Full resolution order
```
