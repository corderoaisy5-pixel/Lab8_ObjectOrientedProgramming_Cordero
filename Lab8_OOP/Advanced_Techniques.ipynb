```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Woof!"

class Cat(Animal):
    def sound(self):
        return "Meow!"

dog = Dog()
cat = Cat()
print(dog.sound())   # Woof!
print(cat.sound())   # Meow!
```

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    course: str

s1 = Student("Aisy", 19, "MECHANICAL")
s2 = Student("Pat", 18, "PSYCHOLOGY")
s3 = Student("Lira", 19, "NURSING")

print(s1)           # Student(name='Aisy', age=19, course='MECHANICAL')
print(s1 == s3)     # True  — __eq__ is auto-generated
print(s1 == s2)     # False
```

```python
class InvalidAgeError(Exception):
    pass

class NegativeScoreError(Exception):
    pass

def set_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")
    print(f"Age set to: {age}")

def set_score(score):
    if score < 0:
        raise NegativeScoreError("Score cannot be negative.")
    print(f"Score set to: {score}")

# Test valid inputs
set_age(20)
set_score(95)

# Test invalid inputs
try:
    set_age(-5)
except InvalidAgeError as e:
    print(f"InvalidAgeError: {e}")

try:
    set_score(-10)
except NegativeScoreError as e:
    print(f"NegativeScoreError: {e}")
```

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero!")
        self._celsius = value

    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32

temp = Temperature(100)
print(f"Celsius:    {temp.celsius}°C")
print(f"Fahrenheit: {temp.fahrenheit}°F")

temp.celsius = 0                          # uses the setter
print(f"Updated:    {temp.celsius}°C = {temp.fahrenheit}°F")

try:
    temp.celsius = -300                   # triggers ValueError
except ValueError as e:
    print(f"ValueError: {e}")
```

```python
class Engine:
    def start(self):
        print("Engine started.")

    def stop(self):
        print("Engine stopped.")

class GPS:
    def navigate(self, destination):
        print(f"Navigating to {destination}...")

class Car:
    def __init__(self):
        self.engine = Engine()    # Car HAS-AN Engine
        self.gps = GPS()          # Car HAS-A GPS

    def drive(self, destination):
        self.engine.start()
        self.gps.navigate(destination)
        print("Car is moving.")

    def park(self):
        self.engine.stop()
        print("Car is parked.")

my_car = Car()
my_car.drive("SM Mall of Asia")
my_car.park()
```
